```python
import numpy as np

# création de la liste de toutes les matières (ressources et SAE)
matière=[f"R{i}" for i in range (101,116)]+[f"SAE{i}" for i in range(11,17)]


"""
# récupération des notes de toutes les matières
note=[float(input(f"Quelle est la note pour {elt} ?")) for elt in matière]
"""

# c'est pour générer des notes allant de 0 à 20 : 
note=[i for i in range(21)]

# création des listes de coéfficients
coeff1=[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coeff2=[4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coeff3=[4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

# round c'est pour les nombres après la virgule, averrage c'est pour la moyenne, weights c'est pour les poids
def moy(coeff):
    moy=round(np.average(note,weights=coeff),3)
    return moy
print (f"Les moyennes des UE sont respectivement UE1 {moy(coeff1)}, UE2 {moy(coeff2)} et UE3 {moy(coeff3)}")
```
